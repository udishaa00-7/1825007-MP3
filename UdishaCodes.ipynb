Write a Python program to calculate the Discrete Fourier Transform (DFT) of the sequence:
\[
x=[1,2,3,4]
\]

```python
import numpy as np
x=[1,2,3,4]
def dft(x):
  N=len(x)
  X=[]
  for k in range(N):
    total=0
    for n in range(N):
      angle= (-2 * (np.pi) * n * k)/N
      total += x[n] * np.exp(1j * angle)
    X.append(total)
  return X
  result=dft(x)
print(np.round(result, 4))
```

Visualisation of mappings of some elementary complex functions w = f(z) from z-plane to w-plane

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([-2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2])
y = np.array([-2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2])

z = x + 1j*y

z = x + y*1j
w = z ** 2

u=np.real(w)
v=np.imag(w)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.title("Z-Plane")
plt.scatter(x,y)
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid(True)

plt.subplot(1,2,2)
plt.title("W-Plane")
plt.scatter(u,v)
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid(True)
plt.show()

```

Solving improper integrals over entire real axis or the positive real axis using Gauss Lagaurre and Gauss Hermite quadrature method. Comparison of results with the ones obtained by contour integration analytically

```python
import numpy as np

n=150
x1,w1 = np.polynomial.laguerre.laggauss(n)
x2,w2 = np.polynomial.hermite.hermgauss(n)


func1 = np.exp(x1)/(1 + x1**2)
product1 = w1 * func1
func2 = np.exp(x2**2)/(1 + x2**2)
product2 = w2 * func2

I1 = np.sum(product1)
I2 = np.sum(product2)
print("Value of improper integral via Gauss Laguerre",I1)
print("Value of improper integral via Gauss Hermite",I2)

```

Comparison of convergence of improper integral computed by Newton Cotes and Gauss
Quadrature Methods.

```python

import numpy as np
from scipy.integrate import simpson

def f(x):
    return 1/((1-x)**2 + x**2)

exact = np.pi/2

for n in [4, 8, 16, 32]:

    # Newton-Cotes
    x = np.linspace(0, 1, n+1)
    I1 = simpson(f(x), x=x)

    # Gauss Quadrature
    x, w = np.polynomial.legendre.leggauss(n)
    I2 = np.sum(w*f((x+1)/2))/2

    print("n =", n)
    print("Newton Cotes =", I1)
    print("Gauss Quadrature =", I2)
    print()

print("Exact value =", exact)
```

The equilibrium temperature of a bar of length L with insulated horizontal sides and the
ends maintained at fixed temperatures.

```python

import numpy as np
import matplotlib.pyplot as plt

L = 10
T1 = 100
T2 = 50

x = np.linspace(0, L, 50)

T = T1 + (T2-T1)*x/L

plt.plot(x, T)
plt.xlabel("Length of bar")
plt.ylabel("Temperature")
plt.title("Equilibrium Temperature Distribution")
plt.grid()
plt.show()
```

Solve a BVP using the Shooting Method:
y''=-2, y(0)=0, y(1)=0
